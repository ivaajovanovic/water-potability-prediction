```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('water_potability.csv')
```

```python
# filling null values
df['ph'] = df.groupby('Potability')['ph'].transform(lambda x: x.fillna(x.mean()))
df['Sulfate'] = df.groupby('Potability')['Sulfate'].transform(lambda x: x.fillna(x.mean()))
df['Trihalomethanes'] = df.groupby('Potability')['Trihalomethanes'].transform(lambda x: x.fillna(x.mean()))
```

```python
# handling outliers - capping method
feature_cols = df.drop('Potability', axis=1).columns
for col in feature_cols:
    lower_bound = df[col].quantile(0.01)
    upper_bound = df[col].quantile(0.99)
    df[col] = np.clip(df[col], lower_bound, upper_bound)
```

```python
print("Descriptive statistics AFTER handling outliers:")
display(df.describe())
```

```python
output_filename = 'water_potability_clean.csv'
df.to_csv(output_filename, index=False)
```
